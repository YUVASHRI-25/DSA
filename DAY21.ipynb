```python
def getNext(n: int) -> int:
    d = 0
    while n > 0:
        e = n % 10
        d += e * e
        n //= 10
    return d

def isHappy(n: int) -> bool:
    a = n
    b = getNext(n)

    while b != 1 and a != b:
        a = getNext(a)
        b = getNext(getNext(b))
    
    return b == 1

n = int(input())
if isHappy(n):
    print(n, "is a Happy Number")
else:
    print(n, "is NOT a Happy Number ")
```
